# ASL Model Converter (Keras 3 to ONNX)
This notebook converts your `.keras` model to `.onnx` for use in the Vibe App.

### Instructions:
1. Run the **Installation** cell below.
2. Run the **Upload** cell and select your `asl_model.keras` file.
3. Run the **Conversion** cell.
4. Download the resulting `asl_model.onnx` and place it in `vibe/public/models/`.

```python
from google.colab import files
import os

print("Upload your asl_model.keras file:")
uploaded = files.upload()
input_model = list(uploaded.keys())[0]
os.rename(input_model, 'asl_model.keras')
print(f"File '{input_model}' uploaded and renamed to 'asl_model.keras'")
```

```python
import tensorflow as tf
import tf2onnx

print("Loading Keras model...")
model = tf.keras.models.load_model('asl_model.keras')

print("Converting to ONNX...")
input_signature = [tf.TensorSpec(shape=(None, 224, 224, 3), dtype=tf.float32, name='input')]
onnx_model, _ = tf2onnx.convert.from_keras(model, input_signature=input_signature, opset=17)

with open("asl_model.onnx", "wb") as f:
    f.write(onnx_model.SerializeToString())

print("✓ Conversion complete: asl_model.onnx")
files.download('asl_model.onnx')
```
